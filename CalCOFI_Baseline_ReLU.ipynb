```python
from pathlib import Path

import numpy as np
import pandas as pd

SEED = 11
RUN_MODE = "predict"
VALIDATION_SEED = 42
REFIT_EPOCHS = 328
RESELECT_EPOCHS = False
OUTPUT_FILE = Path("submission_baseline_relu.csv")
```

```python
train=pd.read_csv("/kaggle/input/datasets/user/competition1/train.csv")
test=pd.read_csv("/kaggle/input/datasets/user/competition1/test.csv")
sub=pd.read_csv("/kaggle/input/datasets/user/competition1/sample_submission.csv")
```

```python
import torch
import torch.nn as nn
from sklearn.model_selection import KFold, train_test_split
from sklearn.preprocessing import StandardScaler

torch.set_num_threads(1)
torch.use_deterministic_algorithms(True)

FEATURES = [
    "Lat_Dec", "Lon_Dec", "NO2uM", "NO3uM", "NH3uM",
    "R_TEMP", "R_Depth", "R_Sal", "R_DYNHT", "R_Nuts",
    "R_Oxy_micromol.Kg", "PO4uM", "SiO3uM", "Salinity1", "Temperature_degC",
]
X = train[FEATURES].to_numpy(dtype=float)
y = train["DIC"].to_numpy(dtype=float)
X_test = test[FEATURES].to_numpy(dtype=float)


class Preprocessor:
    def __init__(self, X_fit, y_fit):
        self.fill_values = np.nanmean(X_fit, axis=0)
        filled = np.where(np.isnan(X_fit), self.fill_values, X_fit)
        self.scaler = StandardScaler().fit(filled)
        self.y_mean = float(y_fit.mean())
        self.y_scale = float(y_fit.std())
        if not np.isfinite(self.y_scale) or self.y_scale == 0:
            raise ValueError("Training DIC must have a finite, nonzero standard deviation")

    def transform_x(self, X_eval):
        filled = np.where(np.isnan(X_eval), self.fill_values, X_eval)
        return self.scaler.transform(filled).astype(np.float32)

    def transform_y(self, y_fit):
        return ((y_fit - self.y_mean) / self.y_scale).astype(np.float32)

    def inverse_y(self, scaled):
        return np.asarray(scaled, dtype=np.float64) * self.y_scale + self.y_mean


if RUN_MODE == "validate":
    X_fit, X_val, y_fit, y_val = train_test_split(
        X, y, test_size=0.2, random_state=VALIDATION_SEED
    )
elif RUN_MODE == "predict":
    X_fit, y_fit = X, y
else:
    raise ValueError('RUN_MODE must be "predict" or "validate"')

prep = Preprocessor(X_fit, y_fit)
X_train = prep.transform_x(X_fit)
y_train = prep.transform_y(y_fit)
if RUN_MODE == "validate":
    X_val_scaled = prep.transform_x(X_val)
else:
    test_scaled = prep.transform_x(X_test)
```

```python
from torch.utils.data import Dataset, DataLoader

class OceanChemistryDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]


def make_train_loader(X_train, y_train):
    train_dataset = OceanChemistryDataset(X_train, y_train)
    return DataLoader(train_dataset, batch_size=32, shuffle=True)


train_loader = make_train_loader(X_train, y_train)
```

```python
class MLPModel(nn.Module):
    def __init__(self, input_size):
        super(MLPModel, self).__init__()
        self.fc1 = nn.Linear(input_size, 128)
        self.relu1 = nn.ReLU()
        self.fc2 = nn.Linear(128, 64)
        self.relu2 = nn.ReLU()
        self.fc3 = nn.Linear(64, 1)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu1(x)
        x = self.fc2(x)
        x = self.relu2(x)
        x = self.fc3(x)
        return x


torch.manual_seed(SEED)
model = MLPModel(input_size=X_train.shape[1])
print(X_train.shape)
```

```python
import copy
import torch.optim as optim

def train_model(model, train_loader, epochs, validation=None):
    target_scale = prep.y_scale if validation is None else validation[2].y_scale
    criterion = nn.HuberLoss(delta=6.0 / target_scale)
        
    optimizer = optim.Adam(model.parameters(), lr=0.003, weight_decay=0.0001)
    scheduler = optim.lr_scheduler.MultiStepLR(
        optimizer, milestones=[300, 700, 1100], gamma=0.3
    )
    best_score = float("inf")
    best_epoch = epochs
    best_weights = None

    for epoch in range(1, epochs + 1):
        model.train()
        for X_batch, y_batch in train_loader:
            optimizer.zero_grad()
            outputs = model(X_batch).squeeze(-1)
            loss = criterion(outputs, y_batch)
            if not torch.isfinite(loss):
                raise FloatingPointError(f"Nonfinite loss at epoch {epoch}")
            loss.backward()
            optimizer.step()

        if validation is not None:
            val_tensor, y_val, val_prep = validation
            model.eval()
            with torch.no_grad():
                outputs = model(val_tensor).squeeze(-1).numpy()
            predictions = val_prep.inverse_y(outputs)
            score = float(np.sqrt(np.mean((predictions - y_val) ** 2)))
            if score < best_score - 1e-5:
                best_score = score
                best_epoch = epoch
                best_weights = copy.deepcopy(model.state_dict())

        scheduler.step()
        if validation is not None and epoch - best_epoch >= 220:
            break

    if best_weights is not None:
        model.load_state_dict(best_weights)
    return best_epoch


def select_epochs(X_fit, y_fit):
    splitter = KFold(n_splits=3, shuffle=True, random_state=123)
    best_epochs = []
    for fold, (fit_rows, val_rows) in enumerate(splitter.split(X_fit), start=1):
        fold_X, fold_y = X_fit[fit_rows], y_fit[fit_rows]
        fold_prep = Preprocessor(fold_X, fold_y)
        fold_loader = make_train_loader(
            fold_prep.transform_x(fold_X), fold_prep.transform_y(fold_y)
        )
        val_tensor = torch.from_numpy(fold_prep.transform_x(X_fit[val_rows]))
        for offset in (100, 200):
            torch.manual_seed(offset + fold)
            fold_model = MLPModel(input_size=fold_X.shape[1])
            best_epochs.append(train_model(
                fold_model, fold_loader, epochs=1600,
                validation=(val_tensor, y_fit[val_rows], fold_prep),
            ))
    return int(np.ceil(np.median(best_epochs)))


if RESELECT_EPOCHS:
    training_rng_state = torch.get_rng_state()
    epochs = select_epochs(X_fit, y_fit)
    torch.set_rng_state(training_rng_state)
else:
    epochs = REFIT_EPOCHS

train_model(model, train_loader, epochs=epochs)
```

```python
model.eval()

if RUN_MODE == "validate":
    val_tensor = torch.tensor(X_val_scaled, dtype=torch.float32)
    with torch.no_grad():
        validation_prediction = prep.inverse_y(model(val_tensor).squeeze(-1).numpy())
    validation_rmse = np.sqrt(np.mean((validation_prediction - y_val) ** 2))
    print(f"Validation RMSE ({len(y_val)} rows, split seed {VALIDATION_SEED}): {validation_rmse:.4f}")
else:
    test_tensor = torch.tensor(test_scaled, dtype=torch.float32)
    with torch.no_grad():
        test_prediction = prep.inverse_y(model(test_tensor).squeeze(-1).numpy())

    if test_prediction.shape != (len(test),) or not np.isfinite(test_prediction).all():
        raise ValueError("The test predictions are incomplete or nonfinite")

    submission = pd.DataFrame({"id": test["id"], "DIC": test_prediction})
    submission.to_csv(OUTPUT_FILE, index=False)
    print(f"Saved {len(submission)} predictions to {OUTPUT_FILE}")
```
